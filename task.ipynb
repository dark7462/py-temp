```python
marks = [
    [10,20,30],
    [40,50,60],
    [70,80,90],
    [1,2,3]
]
for i in marks:
    for j in i: # for-each for for-each loop
        print(j, end = " ")
    print()

for a , b, c in marks: # can use this for definite number of inner iterables
    print(a,b,c)

for i in range(len(marks)): # by using proper index
    for j in range(len(marks[i])):
        print(marks[i][j], end = " ")
    print()
```

```python
# pop() remove the value in the given index , if no element is passed then last element will be removed
# remove () remove the first occurance of the value
lst = [1020,30,50,20,30]
print(lst)
lst.pop() # removes the last element from the list
print(lst)
lst.pop(0) # removes the element in the index
print(lst)
lst.remove(50) # removes the first 50 found
print(lst)
```

```python
# reverse() , sort()
lst = [1020,30,50,20,30]

lst.sort() # sorts the array in O(N logN)
print(lst) # lst.sort(reverse = true) this will sort in reverse order

lst.reverse() # reverse the list in O(N)
print(lst)
```
